# Data Cleaning #

## Handling missing values in Matches dataset ##

```python
# these 2 columns record how many scores each team (home or away) has for penalty shootout
# which does not happen when a match has a winner by the end of time 
# We cannot just drop them or fill in with anything rather than a zero otherwise it would skew the dataset 

df_matches_cleaned['home_penalty_score'] = df_matches_cleaned['home_penalty_score'].fillna(0).astype(int)
df_matches_cleaned['away_penalty_score'] = df_matches_cleaned['away_penalty_score'].fillna(0).astype(int)

# verify to check there is no NaN left
display(df_matches_cleaned.isnull().sum())
```

# Data Aggregation and Analysis #
## Team performance comparison based on different metrics (goals scored, possession, card distribution etc) ##

```python
# combine the relavant datasets into one
df_obj_A = pd.merge(df_team_stats_cleaned, df_teams_cleaned[['team_id', 'team_name']], on='team_id', how='inner')
df_obj_A = pd.merge(df_obj_A, df_matches_cleaned[['match_id', 'result_type']], on='match_id', how='inner')

# calculate the average per match for each team 
team_avg = df_obj_A.groupby('team_name')[
    ['possession_pct', 'shots_on_target', 'fouls', 'saves']
].mean().reset_index()

display(team_avg.head())
display(team_avg.tail())

# possession control
best_possession = team_avg.loc[team_avg['possession_pct'].idxmax()]
worst_possession = team_avg.loc[team_avg['possession_pct'].idxmin()]

# attacking threat
best_attack = team_avg.loc[team_avg['shots_on_target'].idxmax()]
worst_attack = team_avg.loc[team_avg['shots_on_target'].idxmin()]

# discipline
most_fouls = team_avg.loc[team_avg['fouls'].idxmax()]
cleanest = team_avg.loc[team_avg['fouls'].idxmin()]

# defensive  
most_saves = team_avg.loc[team_avg['saves'].idxmax()]
fewest_saves = team_avg.loc[team_avg['saves'].idxmin()]

print("Team Performance")
print(f"- Possession: Highest = {best_possession['team_name']} ({best_possession['possession_pct']}%) | Lowest = {worst_possession['team_name']} ({worst_possession['possession_pct']}%)")
print(f"- Attack: Highest = {best_attack['team_name']} ({best_attack['shots_on_target']}) | Lowest = {worst_attack['team_name']} ({worst_attack['shots_on_target']})")
print(f"- Discipline: Most Fouls = {most_fouls['team_name']} ({most_fouls['fouls']}) | Fewest = {cleanest['team_name']} ({cleanest['fouls']})")
print(f"- Saves: Highest = {most_saves['team_name']} ({most_saves['saves']}) | Lowest = {fewest_saves['team_name']} ({fewest_saves['saves']})")
```

## Player and squad market valuation analysis ##

```python
# merge player data with team names
df_squad_valuation = pd.merge(df_squads_players_cleaned, df_teams_cleaned[['team_id', 'team_name']], on='team_id', how='inner')

# calculate the Total Squad Value 
squad_financials = df_squad_valuation.groupby('team_name')[['market_value_eur']].sum().reset_index()
squad_financials = squad_financials.rename(columns={'market_value_eur': 'total_value'})

# calculate the Average Player Value 
squad_avg = df_squad_valuation.groupby('team_name')[['market_value_eur']].mean().reset_index()

# add the averages into main financials table
squad_financials['avg_player_value'] = squad_avg['market_value_eur']

# the most and least valuable squads
most_valuable_squad = squad_financials.loc[squad_financials['total_value'].idxmax()]
least_valuable_squad = squad_financials.loc[squad_financials['total_value'].idxmin()]

# the single most valuable player in the tournament
most_valuable_player = df_squad_valuation.loc[df_squad_valuation['market_value_eur'].idxmax()]

print("Market Valuation")
print(f"- Total Squad Value: Highest = {most_valuable_squad['team_name']} (€{most_valuable_squad['total_value']}) | Lowest = {least_valuable_squad['team_name']} (€{least_valuable_squad['total_value']})")
print(f"- Avg Player Value: Highest = {most_valuable_squad['team_name']} (€{most_valuable_squad['avg_player_value']}) | Lowest = {least_valuable_squad['team_name']} (€{least_valuable_squad['avg_player_value']})")
print(f"- Most Valuable Player: {most_valuable_player['player_name']} ({most_valuable_player['team_name']}) at €{most_valuable_player['market_value_eur']}")

print("\n Top 5 Most Valuable Squads")
display(squad_financials.sort_values(by='total_value', ascending=False).head())
```

## How do stadium elevation and location affect match tempo or fatigue? ##

```python
import requests

weather_records = []

# get elevation and typical tournament weather for each stadium
for index, venue in df_venues_cleaned.iterrows():
    lat = venue['latitude']
    lon = venue['longitude']
    venue_id = venue['venue_id']

    # pulling elevation and forecast/daily temperatures from Open Meteo API
    url = f"https://api.open-meteo.com/v1/forecast?latitude={lat}&longitude={lon}&daily=temperature_2m_max,apparent_temperature_max&timezone=auto"
    
    response = requests.get(url)
    data = response.json()
    
    # extract elevation from the API 
    api_elevation = data.get('elevation')
    
    # extract average daily max and feels-like temperature
    daily_temps = data['daily']['temperature_2m_max']
    daily_feels_like = data['daily']['apparent_temperature_max']
    
    avg_temp = sum(daily_temps) / len(daily_temps)
    avg_feels_like = sum(daily_feels_like) / len(daily_feels_like)
    
    # store in records list
    weather_records.append({
        'venue_id': venue_id,
        'elevation_m': api_elevation,
        'avg_temp': avg_temp,
        'avg_feels_like': avg_feels_like
    })

# convert the API data to a dataframe
df_open_meteo = pd.DataFrame(weather_records)

print("Open-Meteo API Results")
display(df_open_meteo.head())
```

```python
# add to venues with Open Meteo data
df_venues_plus = pd.merge(df_venues_cleaned, df_open_meteo, on='venue_id', how='inner')

# link matches to venues
df_matches_env = pd.merge(df_matches_cleaned[['match_id', 'venue_id']], df_venues_plus, on='venue_id', how='inner')

# combine with team performance metrics (tempo = shots, fatigue = fouls)
df_obj_C = pd.merge(df_team_stats_cleaned, df_matches_env, on='match_id', how='inner')

# group by venue to calculate average match attributes with environmental factors
venue_summary = df_obj_C.groupby(['stadium_name', 'city', 'elevation_m', 'avg_temp', 'avg_feels_like'])[
    ['total_shots', 'shots_on_target', 'fouls']
].mean().reset_index()

print("Venue Summary Dataset (Venues + Weather + Match Stats)")
display(venue_summary.head())
```

```python
# elevation extremes
highest_alt = venue_summary.loc[venue_summary['elevation_m'].idxmax()]
lowest_alt = venue_summary.loc[venue_summary['elevation_m'].idxmin()]

# temperature extremes
hottest_venue = venue_summary.loc[venue_summary['avg_temp'].idxmax()]
coldest_venue = venue_summary.loc[venue_summary['avg_temp'].idxmin()]

# tempo and fatigue extremes
highest_tempo = venue_summary.loc[venue_summary['total_shots'].idxmax()]
lowest_tempo = venue_summary.loc[venue_summary['total_shots'].idxmin()]
most_fouls = venue_summary.loc[venue_summary['fouls'].idxmax()]
cleanest_venue = venue_summary.loc[venue_summary['fouls'].idxmin()]

print("Environmental & Match Extremes\n")

print(f"Elevation: Highest = {highest_alt['stadium_name']} ({highest_alt['elevation_m']}m) | Lowest = {lowest_alt['stadium_name']} ({lowest_alt['elevation_m']}m)")
print(f"Climate: Hottest = {hottest_venue['stadium_name']} ({hottest_venue['avg_temp']}°C) | Coolest = {coldest_venue['stadium_name']} ({coldest_venue['avg_temp']}°C)")
print(f"Match tempo: Highest = {highest_tempo['stadium_name']} ({highest_tempo['total_shots']} shots/game) | Lowest = {lowest_tempo['stadium_name']} ({lowest_tempo['total_shots']} shots/game)")
print(f"Fatigue indicator: Most Fouls = {most_fouls['stadium_name']} ({most_fouls['fouls']} fouls/game) | Fewest = {cleanest_venue['stadium_name']} ({cleanest_venue['fouls']} fouls/game)")
```
