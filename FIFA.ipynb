## How do stadium elevation and location affect match tempo or fatigue? ##

```python
import requests

weather_records = []

# get elevation and typical tournament weather for each stadium
for index, venue in df_venues_cleaned.iterrows():
    lat = venue['latitude']
    lon = venue['longitude']
    venue_id = venue['venue_id']

    # pulling elevation and forecast/daily temperatures from Open Meteo API
    url = f"https://api.open-meteo.com/v1/forecast?latitude={lat}&longitude={lon}&daily=temperature_2m_max,apparent_temperature_max&timezone=auto"
    
    response = requests.get(url)
    data = response.json()
    
    # extract elevation from the API 
    api_elevation = data.get('elevation')
    
    # extract average daily max and feels-like temperature
    daily_temps = data['daily']['temperature_2m_max']
    daily_feels_like = data['daily']['apparent_temperature_max']
    
    avg_temp = sum(daily_temps) / len(daily_temps)
    avg_feels_like = sum(daily_feels_like) / len(daily_feels_like)
    
    # store in records list
    weather_records.append({
        'venue_id': venue_id,
        'elevation_m': api_elevation,
        'avg_temp': avg_temp,
        'avg_feels_like': avg_feels_like
    })

# convert the API data to a dataframe
df_open_meteo = pd.DataFrame(weather_records)

print("Open-Meteo API Results")
display(df_open_meteo.head())
```

```python
# add to venues with Open Meteo data
df_venues_plus = pd.merge(df_venues_cleaned, df_open_meteo, on='venue_id', how='inner')

# link matches to venues
df_matches_env = pd.merge(df_matches_cleaned[['match_id', 'venue_id']], df_venues_plus, on='venue_id', how='inner')

# combine with team performance metrics (tempo = shots, fatigue = fouls)
df_obj_C = pd.merge(df_team_stats_cleaned, df_matches_env, on='match_id', how='inner')

# group by venue to calculate average match attributes with environmental factors
venue_summary = df_obj_C.groupby(['stadium_name', 'city', 'elevation_m', 'avg_temp', 'avg_feels_like'])[
    ['total_shots', 'shots_on_target', 'fouls']
].mean().reset_index()

print("Venue Summary Dataset (Venues + Weather + Match Stats)")
display(venue_summary.head())
```

```python
# elevation extremes
highest_alt = venue_summary.loc[venue_summary['elevation_m'].idxmax()]
lowest_alt = venue_summary.loc[venue_summary['elevation_m'].idxmin()]

# temperature extremes
hottest_venue = venue_summary.loc[venue_summary['avg_temp'].idxmax()]
coldest_venue = venue_summary.loc[venue_summary['avg_temp'].idxmin()]

# tempo and fatigue extremes
highest_tempo = venue_summary.loc[venue_summary['total_shots'].idxmax()]
lowest_tempo = venue_summary.loc[venue_summary['total_shots'].idxmin()]
most_fouls = venue_summary.loc[venue_summary['fouls'].idxmax()]
cleanest_venue = venue_summary.loc[venue_summary['fouls'].idxmin()]

print("Environmental & Match Extremes\n")

print(f"Elevation: Highest = {highest_alt['stadium_name']} ({highest_alt['elevation_m']}m) | Lowest = {lowest_alt['stadium_name']} ({lowest_alt['elevation_m']}m)")
print(f"Climate: Hottest = {hottest_venue['stadium_name']} ({hottest_venue['avg_temp']}°C) | Coolest = {coldest_venue['stadium_name']} ({coldest_venue['avg_temp']}°C)")
print(f"Match tempo: Highest = {highest_tempo['stadium_name']} ({highest_tempo['total_shots']} shots/game) | Lowest = {lowest_tempo['stadium_name']} ({lowest_tempo['total_shots']} shots/game)")
print(f"Fatigue indicator: Most Fouls = {most_fouls['stadium_name']} ({most_fouls['fouls']} fouls/game) | Fewest = {cleanest_venue['stadium_name']} ({cleanest_venue['fouls']} fouls/game)")
```

## Association of match outcomes using FIFA rankings, rolling form, and team strength etc ##

```python
# merge the home team's stats to the matches
df_home_merged = pd.merge(
    df_matches_cleaned,
    df_teams_cleaned[['team_id', 'team_name', 'fifa_ranking_pre_tournament', 'elo_rating']],
    left_on='home_team_id',
    right_on='team_id',
    how='inner'
).rename(columns={
    'team_name': 'home_team',
    'fifa_ranking_pre_tournament': 'home_fifa_rank',
    'elo_rating': 'home_elo'
}).drop(columns=['team_id'])

display(df_home_merged)
```

```python
# merge the away team's stats to the same table
df_obj_D = pd.merge(
    df_home_merged,
    df_teams_cleaned[['team_id', 'team_name', 'fifa_ranking_pre_tournament', 'elo_rating']],
    left_on='away_team_id',
    right_on='team_id',
    how='inner'
).rename(columns={
    'team_name': 'away_team',
    'fifa_ranking_pre_tournament': 'away_fifa_rank',
    'elo_rating': 'away_elo'
}).drop(columns=['team_id'])

display(df_obj_D)
```

```python
# create a Python function to find out the winner
def get_winner(row):
    if row['home_score'] > row['away_score']:
        return 'Home Win'
    elif row['home_score'] < row['away_score']:
        return 'Away Win'
    else:
        return 'Draw'

# apply the function to create a new column
df_obj_D['winner'] = df_obj_D.apply(get_winner, axis=1)

# create a function for the FIFA rankings (lower number is better)
def get_fifa(row):
    if row['home_fifa_rank'] < row['away_fifa_rank']:
        return 'Home Win'
    elif row['home_fifa_rank'] > row['away_fifa_rank']:
        return 'Away Win'
    else:
        return 'Equal'

# apply the function
df_obj_D['fifa_favorite'] = df_obj_D.apply(get_fifa, axis=1)

# create a function for the Elo ratings (higher number is better)
def get_elo(row):
    if row['home_elo'] > row['away_elo']:
        return 'Home Win'
    elif row['home_elo'] < row['away_elo']:
        return 'Away Win'
    else:
        return 'Equal'

# apply it
df_obj_D['elo_favorite'] = df_obj_D.apply(get_elo, axis=1)

display(df_obj_D[['home_team', 'away_team', 'winner', 'fifa_favorite', 'elo_favorite']].head())
```

```python
# check where the prediction matched reality
fifa_correct = df_obj_D['winner'] == df_obj_D['fifa_favorite']
elo_correct = df_obj_D['winner'] == df_obj_D['elo_favorite']

# calculate the percentage 
fifa_accuracy = fifa_correct.mean() * 100
elo_accuracy = elo_correct.mean() * 100

print("Predictive Power of FIFA Rank vs Elo Rating\n")
print(f"Total Matches Analyzed: {len(df_obj_D)}")
print(f"FIFA Ranking Accuracy: {fifa_accuracy}%")
print(f"Elo Rating Accuracy: {elo_accuracy}%")
```

### VISUALIZATIONS 2 ###

```python
# get a list of every unique team in the tournament
all_teams = pd.concat([df_obj_D['home_team'], df_obj_D['away_team']]).unique()

# calculate the accuracy for each individual team
team_data = []
for team in all_teams:
    
    team_filter = df_obj_D[(df_obj_D['home_team'] == team) | (df_obj_D['away_team'] == team)]
    
    # calculate how often the rankings were right for this team
    fifa_acc = (team_filter['winner'] == team_filter['fifa_favorite']).mean() * 100
    elo_acc = (team_filter['winner'] == team_filter['elo_favorite']).mean() * 100
    
    # save to  list
    team_data.append({
        'Team': team,
        'FIFA Accuracy': fifa_acc,
        'Elo Accuracy': elo_acc
    })

# convert the list into a data frame
team_acc_df = pd.DataFrame(team_data)

plot = sns.regplot(
    data=team_acc_df, 
    x='FIFA Accuracy', 
    y='Elo Accuracy', 
    scatter_kws={'s': 70, 'color': 'purple'}, 
    line_kws={'color': 'red', 'linestyle': '-'} 
)
```
